# Necessary Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

# Part A — Uniform Distribution

```python
np.random.seed(42)
waiting_times = np.random.uniform(low=3, high=15, size=1000)
mean_val=np.mean(waiting_times)
median_val=np.median(waiting_times)
dev_val=np.std(waiting_times)
print("First 10 waiting times:",waiting_times[:10])
print("Mean: ",mean_val)
print("Median: ",median_val)
print("Standard Deviation: ",dev_val)
```

#  Part B — Normal Distribution

```python
wait_time_raw = np.random.normal(loc=8, scale=2, size=3000)
wait_time_b =np.maximum(0,wait_time_raw)
mean_val_b=np.mean(wait_time_b)
median_val_b=np.median(wait_time_b)
dev_val_b=np.std(wait_time_b)
max_val=np.max(wait_time_b)
print("Mean: ",mean_val_b)
print("Median: ",median_val_b)
print("Standard Deviation: ",dev_val_b)
print("Max val: ",max_val)
```

# Part C — Visualization

```python
plt.figure(figsize=(12, 5))


# Subplot 1: (Normal Distribution)
plt.subplot(1, 2, 1)
plt.hist(waiting_times, bins=30, edgecolor='black', alpha=0.7, color='lightcoral')
plt.title('Subplot 1(Normal Distribution)*-*')
plt.xlabel('Wait Time (minutes)')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)

# Subplot 2:  (Uniform Distribution)
plt.subplot(1, 2, 2)
plt.hist(wait_time_b, bins=30, edgecolor='black', alpha=0.7, color='skyblue')
plt.title('Subplot 2 (Uniform Distribution)*-*')
plt.xlabel('Wait Time (minutes)')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)

plt.tight_layout()
plt.show()
```
