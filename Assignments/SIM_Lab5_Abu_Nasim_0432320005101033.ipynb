## Problem Statement

**Perform a two-sample t-test on the following data:**

* **Data 1:** \[18.4, 19.1, 17.9, 18.7, 18.3]
* **Data 2:** \[20.2, 20.5, 20.1, 20.3, 20.4]

Use significance level **α = 0.05** to determine whether the population means differ.

### Tasks

1. State $H_0$ and $H_1$. You will assume **equal variances** (pooled).
2. Compute the **sample means**, **sample variances (ddof = 1)**, and the **mean difference**.
3. Calculate the **t-statistic**, **degrees of freedom**, and the **two-tailed p-value**.
4. Construct the **95% confidence interval** for the mean difference.
5. **Decision & Interpretation:** Conclude (reject / fail to reject $H_0$) and explain the result **in one or two plain-English sentences**.



---




### Task 1: State $H_0$ and $H_1$

*   **Null Hypothesis ($H_0$):** There is no significant difference between the population means of the two datasets ($\mu_1 = \mu_2$).
*   **Alternative Hypothesis ($H_1$):** There is a significant difference between the population means of the two datasets ($\mu_1 \neq \mu_2$).

We assume equal variances (pooled t-test).

```python
#Nececery Library
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

### Task 2: Define the datasets and compute sample statistics

```python
# Define the new datasets
data1_new = np.array([18.4, 19.1, 17.9, 18.7, 18.3])
data2_new = np.array([20.2, 20.5, 20.1, 20.3, 20.4])

# Compute sample means
mean1_new = np.mean(data1_new)
mean2_new = np.mean(data2_new)

# Compute sample variances (ddof=1 for sample variance)
s1_sq_new = np.var(data1_new, ddof=1)
s2_sq_new = np.var(data2_new, ddof=1)

# Compute mean difference
mean_diff_new = mean1_new - mean2_new

print(f"Sample 1 Mean: {mean1_new:.2f}")
print(f"Sample 2 Mean: {mean2_new:.2f}")
print(f"Sample 1 Variance: {s1_sq_new:.2f}")
print(f"Sample 2 Variance: {s2_sq_new:.2f}")
print(f"Mean Difference (Sample 1 - Sample 2): {mean_diff_new:.2f}")
```

### Task 3: Calculate t-statistic, degrees of freedom, and p-value

```python
# Perform the two-sample t-test assuming equal variances
t_statistic_new, p_value_new = stats.ttest_ind(data1_new, data2_new, equal_var=True)

# Degrees of freedom
n1_new, n2_new = len(data1_new), len(data2_new)
df_new = n1_new + n2_new - 2

print(f"t-Statistic: {t_statistic_new:.3f}")
print(f"Degrees of Freedom: {df_new}")
print(f"p-Value: {p_value_new:.4f}")
```

### Task 4: Construct the 95% confidence interval for the mean difference

```python
# Pooled variance for confidence interval calculation
sp_sq_new = ((n1_new - 1) * s1_sq_new + (n2_new - 1) * s2_sq_new) / (n1_new + n2_new - 2)

# Standard error using pooled variance
std_err_new = np.sqrt(sp_sq_new * (1/n1_new + 1/n2_new))

# 95% confidence interval for the mean difference
ci_new = stats.t.interval(0.95, df_new, loc=mean_diff_new, scale=std_err_new)

print(f"95% Confidence Interval of Mean Difference: [{ci_new[0]:.3f}, {ci_new[1]:.3f}]")
```

### Task 5: Decision & Interpretation

```python
alpha_new = 0.05  # Significance level

if p_value_new < alpha_new:
    print("Decision: Reject the null hypothesis. ")
    print("Interpretation: Based on the t-test results, there is a statistically significant difference between the population means of the two datasets at the 0.05 significance level.")
else:
    print("Decision: Fail to reject the null hypothesis.")
    print("Interpretation: Based on the t-test results, there is no statistically significant difference between the population means of the two datasets at the 0.05 significance level.")
```
