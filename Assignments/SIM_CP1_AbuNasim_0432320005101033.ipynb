```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

# **1: Advanced Scalar & Array Operations**

```python
scalar = 3.7
arr = np.array([5.5, 8.5, 9.5, 3.5, 1.5, 4.5, 5.5, 6.5])

# addition
print(f"addition: {scalar + arr}")
# subtraction
print(f"subtraction: {scalar - arr}")
# multiplication
print(f"multiplication: {scalar * arr}")
# division
print(f"division: {scalar / arr}")
#power
print(f"power: {scalar ** arr}")

#squar root
print(f"squart root: {np.sqrt(arr)}")
#exponential
print(f"exponential: {np.exp(arr)}")
#natural logarithm
print(f"natural logarithm: {np.log(arr)}")
#log base 10
print(f"log base 10: {np.log10(arr)}")

#sin
print(f"sin: {np.sin(arr)}")
#cos
print(f"cos: {np.cos(arr)}")
#tan
print(f"tan: {np.tan(arr)}")
#arcsin
clip_arr=np.clip(arr,-1,1)
print(f"arcsin: {np.arcsin(clip_arr)}")
#arccos
print(f"arccos: {np.arccos(clip_arr)}")
```

# **2. Matrix Mathematical Operations**

```python
matrix=np.array([
    [1.5, -2.7, 3.1],
    [4.0, 5.8, -0.9],
    [6.2, 7.4, 0.1],
    [-0.5, 1.2, 8.8]
])
#sin
print(f"sin: {np.sin(matrix)}")
#cos
print(f"cos: {np.cos(matrix)}")
#tan
print(f"tan: {np.tan(matrix)}")
#arcos
print(f"arccos: {np.arccos(matrix)}")
#arcsin
print(f"arcsin: {np.arcsin(matrix)}")
#arctan
print(f"arctan: {np.arctan(matrix)}")

#exp
print(f"exp: {np.exp(matrix)}")
#log
print(f"log: {np.log(matrix)}")
#log10
print(f"log10: {np.log10(matrix)}")

#absulute value
print(f"absulute value: {np.abs(matrix)}")
#squar root
print(f"squar root: {np.sqrt(matrix)}")
#squar
print(f"squar: {np.square(matrix)}")
#cube root
print(f"cube root: {np.cbrt(matrix)}")

#reminder when div by 3
print(f"reminder when div by 3: {np.remainder(matrix,3)}")
#round
print(f"round: {np.round(matrix)}")
#ceil
print(f"ceil: {np.ceil(matrix)}")
#floor
print(f"floor: {np.floor(matrix)}")
```

# **3. Matrix Statistics and Analysis**

```python
mat_34 = np.random.uniform(-10, 10, (3, 4))
#max value
print(f"max value: {np.max(mat_34)}")
#min value
print(f"min value: {np.min(mat_34)}")
#index position
print(f"index position: {np.argmax(mat_34)}")
#index position
print(f"index position: {np.argmin(mat_34)}")
#sum of all elements
print(f"sum of all elements: {np.sum(mat_34)}")
#mean of all elements
print(f"mean of all elements: {np.mean(mat_34)}")
#product of all elements
print(f"product of all elements: {np.prod(mat_34)}")
#standard deviation
print(f"standard deviation: {np.std(mat_34)}")
#median
print(f"median: {np.median(mat_34)}")
#variance
print(f"variance: {np.var(mat_34)}")

#sum of each row
print(f"sum of each row: {np.sum(mat_34, axis=0)}")
#sum of each column
print(f"sum of each column: {np.sum(mat_34, axis=1)}")
#max value from each row
print(f"max value from each row: {np.max(mat_34, axis=0)}")
#min value from each col
print(f"min value from each col: {np.min(mat_34, axis=1)}")
#
```

# **4. Matrix Sorting and Searching Operations**

```python
mat_44 = np.random.randint(-20, 20, (4, 4))

#sort each row in asc order
print(f"sort each row in asc order: {np.sort(mat_44, axis=0)}")
#sort each row in dec order
print(f"sort each row in dec order: {np.sort(mat_44, axis=0)[::-1]}")
#sort mat in asc order
print(f"sort mat in asc order: {np.sort(mat_44)}")
#sort mat in dec order
print(f"sort mat in dec order: {np.sort(mat_44)[::-1]}")


#five largest value in mat
print(f"five largest value in mat: {np.partition(mat_44.flatten(), -5)[-5:]}")
#five smallest value in mat
print(f"five smallest value in mat: {np.partition(mat_44.flatten(), 5)[:5]}")
#Positions of elements greater than the matrix mean
print(f"Positions of elements greater than the matrix mean: {np.where(mat_44 > np.mean(mat_44))}")
#cnt of pos val
print(f"cnt of pos val: {np.count_nonzero(mat_44 > 0)}")
#cnt of neg val
print(f"cnt of neg val: {np.count_nonzero(mat_44 < 0)}")
#cnt of zero val
print(f"cnt of zero val: {np.count_nonzero(mat_44 == 0)}")
#
```
