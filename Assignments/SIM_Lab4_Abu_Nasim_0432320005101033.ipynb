## Necessary Libraries

```python
import random
import numpy as np
import math
```

## Part 1 — Setup & Personalization

```python
my_id = 33
R = my_id/10

full_name = "Abu Nasim"

# Count letters in name
letter_count = 0
for char in full_name:
    if char.isalpha():
        letter_count = letter_count + 1

# The number of points to generate
base_points = 200 * letter_count

print("Range Used: [0, ", R, "]")
print("Number of letters in name:", letter_count)
print("Starting points:", base_points)
print("")
```

## Part 2 — Simulation Logic

```python
def is_inside(x, y, rad):
  return (x*x)+(y*y)<=rad*rad

def run_simulation(total_points, R):
    half = total_points // 2
    inside_count = 0

    # random.uniform()
    for count in range(half):
        x = random.uniform(0, R)
        y = random.uniform(0, R)
        if is_inside(x, y, R):
            inside_count += 1

    # numpy.random.rand()
    # np.random.rand() gives 0 to 1, so we multiply by R
    x_array = np.random.rand(total_points - half) * R
    y_array = np.random.rand(total_points - half) * R

    for index in range(len(x_array)):
        x = x_array[index]
        y = y_array[index]
        if is_inside(x, y, R):
            inside_count += 1

    return inside_count
```

## Part 3 — Output & Analysis

```python
print(f"{'Range Used'} | {'Total Points'} | {'Points Inside'} | {'Estimated π'} | {'Error from math.pi'}")
print("----------------------------------------------------------------------------")

# Run the simulation 5 times
for i in [1, 2, 3, 4, 5]:
    total_points = base_points * i


    inside_count = run_simulation(total_points, R)

    # Calculate Pi
    estimated_pi = 4 * (inside_count / total_points)
    error = abs(math.pi - estimated_pi)

    # Print the results for this run
    print(f"{R} | {total_points} | {inside_count} | {estimated_pi} | {error}")
```
